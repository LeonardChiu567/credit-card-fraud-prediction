Amount is very right skewed, median about $22 and max is $25,691. This seems to be normal when it comes to credit card fraud. There are also $0 transactions, which are probably card-verification checks which fraudsters use to test stolen cards.

```python
import matplotlib.pyplot as plt

counts = df["Class"].value_counts()
pct = df["Class"].value_counts(normalize=True) * 100

balance = pd.DataFrame({"count": counts, "percent": pct.round(3)})
balance.index = balance.index.map({0: "Legitimate (0)", 1: "Fraud (1)"})
print(balance)
print(f"\nRatio: about 1 fraud for every {counts[0] / counts[1]:.0f} legitimate transactions")

fig, ax = plt.subplots(figsize=(5, 4))
ax.bar(balance.index, balance["count"], color=["#2a6fb0", "#c0392b"])
ax.set_yscale("log")   # log scale, otherwise the fraud bar is invisible
ax.set_ylabel("Count (log scale)")
ax.set_title("Class balance")
for i, n in enumerate(balance["count"]):
    ax.text(i, n, f"{n:,}", ha="center", va="bottom")
plt.show()
```

As expected, the dataset is heavily imbalanced, with only 0.167% of the data being fraud.

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ks_2samp

features = [c for c in df.columns if c != "Class"]
fraud, legit = df[df["Class"] == 1], df[df["Class"] == 0]

# How different is each feature between the classes?
# KS statistic: 0 = identical distributions, 1 = completely separated
sep = pd.DataFrame({
    "legit_median": legit[features].median(),
    "fraud_median": fraud[features].median(),
    "KS_stat": [ks_2samp(fraud[f], legit[f]).statistic for f in features],
}).sort_values("KS_stat", ascending=False).round(3)
display(sep)

# Density plot per feature, ordered from most to least separating.
# common_norm=False scales each class separately, otherwise fraud is invisible.
plot_df = df.assign(Amount=np.log1p(df["Amount"]))   # log scale so Amount is readable
fig, axes = plt.subplots(6, 5, figsize=(20, 20))
for ax, f in zip(axes.flat, sep.index):
    lo = min(fraud[f].quantile(0.01), legit[f].quantile(0.001))
    hi = max(fraud[f].quantile(0.99), legit[f].quantile(0.999))
    sns.kdeplot(data=plot_df, x=f, hue="Class", common_norm=False, fill=True, alpha=0.4,
                palette={0: "#2a6fb0", 1: "#c0392b"}, ax=ax, warn_singular=False)
    if f != "Amount":
        ax.set_xlim(lo, hi)   # trim extreme tails so the shapes are visible
    ax.set_title(f"{'log(1+Amount)' if f == 'Amount' else f}  (KS = {sep.loc[f, 'KS_stat']:.2f})")
    ax.set_xlabel("")
    ax.set_ylabel("")
fig.suptitle("Feature distributions: legitimate (blue) vs fraud (red)", fontsize=16, y=1.0)
fig.tight_layout()
plt.show()
```

There are certain variables that have a fraud distribution shifted far from non-fraud. 

The time variable is an issue, I will need to convert it to time of day in hours because time currently is seconds from first transaction. By setting it from 0-23 (hours of day, but the exact hour isn't specified in dataset) the model might capture an increase in fraud at certain periods during the day.

```python
df["Hour"] = (df["Time"] // 3600) % 24
df = df.drop(columns="Time")
```

```python
df.head()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

X_vif = add_constant(df.drop(columns="Class"))   # VIF needs an intercept to be computed correctly
vif = pd.DataFrame({
    "feature": X_vif.columns,
    "VIF": [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
})
vif = vif[vif["feature"] != "const"].sort_values("VIF", ascending=False).round(2)
vif
```

Check for multicollinearity for logistic regression baseline, amount has a high VIF value, so we can try log transforming it.

```python

if "Amount" in df.columns:
    df["LogAmount"] = np.log1p(df["Amount"])   # log(1 + x), so $0 amounts stay at 0
    df = df.drop(columns="Amount")

# Recheck VIF
X_vif = add_constant(df.drop(columns="Class"))
vif = pd.DataFrame({
    "feature": X_vif.columns,
    "VIF": [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
})
vif = vif[vif["feature"] != "const"].sort_values("VIF", ascending=False).round(2)
vif
```

For logistic regression, we will do sine/cosine encoding as "hour" isn't a straight line. 

```python
df["Hour_sin"] = np.sin(2 * np.pi * df["Hour"] / 24)
df["Hour_cos"] = np.cos(2 * np.pi * df["Hour"] / 24)
df = df.drop(columns="Hour")
```

We will split into 80/20 training and testing data, making sure to stratify, keeping same fraud rate percentage. 

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns="Class")
y = df["Class"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,      # 80% train, 20% test
    stratify=y,          # keep the same ~0.17% fraud rate in both sets
    random_state=42,     # same split every time you re-run
)

print(f"Train: {len(X_train):,} rows, {y_train.sum()} fraud ({y_train.mean()*100:.3f}%)")
print(f"Test:  {len(X_test):,} rows, {y_test.sum()} fraud ({y_test.mean()*100:.3f}%)")
```

Scale features for logistic regression baseline(Not random forest or XGBoost)

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logreg = Pipeline([
    ("scaler", StandardScaler()),                       # fit on training data only, when the pipeline is fit
    ("model", LogisticRegression(max_iter=1000, random_state=42)),
])
check = StandardScaler().fit(X_train)
X_train_scaled = pd.DataFrame(check.transform(X_train), columns=X_train.columns)
X_train_scaled.describe().loc[["mean", "std"]].T.round(3)
```

Now for the baseline logistic regression model. 

```python
from sklearn.metrics import average_precision_score, precision_score, recall_score, f1_score, confusion_matrix

lr_baseline = logreg.fit(X_train, y_train)            # scaler + LogisticRegression, all defaults

baseline_prob = lr_baseline.predict_proba(X_test)[:, 1]
baseline_pred = lr_baseline.predict(X_test)           # default 0.5 cutoff

print(f"Test PR-AUC:  {average_precision_score(y_test, baseline_prob):.4f}")
print(f"Precision:    {precision_score(y_test, baseline_pred):.3f}")
print(f"Recall:       {recall_score(y_test, baseline_pred):.3f}")
print(f"F1:           {f1_score(y_test, baseline_pred):.3f}")
print(confusion_matrix(y_test, baseline_pred))
```

Now we tune the model. The first cell tries five class weights and keeps the best, the second cell picks a threshold instead of using 0.5, and the final cell tests the final model.

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

lr_grid = GridSearchCV(
    logreg,
    param_grid={"model__class_weight": [None, "balanced", {0: 1, 1: 10}, {0: 1, 1: 50}, {0: 1, 1: 100}]},
    scoring="average_precision",   # PR-AUC
    cv=cv,
    n_jobs=-1,
)
lr_grid.fit(X_train, y_train)

lr_results = pd.DataFrame(lr_grid.cv_results_)[["param_model__class_weight", "mean_test_score", "std_test_score"]]
print(lr_results.round(4).to_string(index=False))
print("Best:", lr_grid.best_params_)
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_recall_curve

lr_best = lr_grid.best_estimator_
lr_oof = cross_val_predict(lr_best, X_train, y_train, cv=cv, method="predict_proba", n_jobs=-1)[:, 1]

prec, rec, thresholds = precision_recall_curve(y_train, lr_oof)
f1 = 2 * prec * rec / (prec + rec + 1e-12)
best_i = np.argmax(f1[:-1])          # the last precision/recall pair has no threshold
lr_threshold = thresholds[best_i]

print(f"Chosen threshold: {lr_threshold:.3f}")
print(f"CV precision={prec[best_i]:.3f}  recall={rec[best_i]:.3f}  F1={f1[best_i]:.3f}")
```

```python
lr_prob = lr_best.predict_proba(X_test)[:, 1]
lr_pred = (lr_prob >= lr_threshold).astype(int)

print(f"Test PR-AUC:  {average_precision_score(y_test, lr_prob):.4f}")
print(f"Precision:    {precision_score(y_test, lr_pred):.3f}")
print(f"Recall:       {recall_score(y_test, lr_pred):.3f}")
print(f"F1:           {f1_score(y_test, lr_pred):.3f}")
print(confusion_matrix(y_test, lr_pred))
```

The PR AUC and Precision is similar, but the recall and F1 are better. The next model will be random forest. 

```python
from sklearn.ensemble import RandomForestClassifier

rf_baseline = RandomForestClassifier(random_state=42, n_jobs=-1)   # all defaults: 100 trees, no class weights
rf_baseline.fit(X_train, y_train)                                  # unscaled data, as trees don't need scaling

rf_base_prob = rf_baseline.predict_proba(X_test)[:, 1]
rf_base_pred = rf_baseline.predict(X_test)                         # default 0.5 cutoff

print(f"Test PR-AUC:  {average_precision_score(y_test, rf_base_prob):.4f}")
print(f"Precision:    {precision_score(y_test, rf_base_pred):.3f}")
print(f"Recall:       {recall_score(y_test, rf_base_pred):.3f}")
print(f"F1:           {f1_score(y_test, rf_base_pred):.3f}")
print(confusion_matrix(y_test, rf_base_pred))
```

The baseline is quite good, now tune hyperparameters by using 1) randomized searching for best hyperparameters 2) finding best cutoff. Cell 3 will use that combination.
The candidate values include default and 10 - 30 max depth, and only 1-10 min_samples per leaf because of imbalanced fraud cases. For the features, there is the default square root, 0.3 and 0.5. Class weights are standard, and max samples are at 0.3 to 0.5 to save time. Number of estimators is fixed at 100.

```python
from sklearn.model_selection import RandomizedSearchCV

cv3 = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

rf_param_dist = {
    "max_depth":        [None, 10, 20, 30],
    "min_samples_leaf": [1, 2, 5, 10],
    "max_features":     ["sqrt", 0.3, 0.5],
    "class_weight":     [None, "balanced", "balanced_subsample"],
    "max_samples":      [0.3, 0.5],
}

rf_search = RandomizedSearchCV(
    RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),   # each forest uses all cores (threads)
    param_distributions=rf_param_dist,
    n_iter=20,                      # 20 random combinations x 3 folds = 60 fits
    scoring="average_precision",    # PR-AUC
    cv=cv3,
    n_jobs=1,                       # run the fits one at a time, so no worker processes are needed
    random_state=42,
)
rf_search.fit(X_train, y_train)

rf_results = pd.DataFrame(rf_search.cv_results_)
cols = [c for c in rf_results.columns if c.startswith("param_")] + ["mean_test_score", "std_test_score"]
display(rf_results.sort_values("mean_test_score", ascending=False)[cols].round(4).head(8))
print("Best:", rf_search.best_params_, "CV PR-AUC:", round(rf_search.best_score_, 4))
```

```python
rf_best = rf_search.best_estimator_.set_params(n_jobs=-1)
rf_oof = cross_val_predict(rf_best, X_train, y_train, cv=cv3, method="predict_proba")[:, 1]

prec, rec, thresholds = precision_recall_curve(y_train, rf_oof)
f1 = 2 * prec * rec / (prec + rec + 1e-12)
best_i = np.argmax(f1[:-1])
rf_threshold = thresholds[best_i]

print(f"Chosen threshold: {rf_threshold:.3f}")
print(f"CV precision={prec[best_i]:.3f}  recall={rec[best_i]:.3f}  F1={f1[best_i]:.3f}")
```

```python
rf_prob = rf_best.predict_proba(X_test)[:, 1]
rf_pred = (rf_prob >= rf_threshold).astype(int)

print(f"Test PR-AUC:  {average_precision_score(y_test, rf_prob):.4f}")
print(f"Precision:    {precision_score(y_test, rf_pred):.3f}")
print(f"Recall:       {recall_score(y_test, rf_pred):.3f}")
print(f"F1:           {f1_score(y_test, rf_pred):.3f}")
print(confusion_matrix(y_test, rf_pred))
```

The tuned random forest outperforms the logistic regression. Last model: XGBoost. Start with a baseline, then tune.

```python
from xgboost import XGBClassifier

xgb_baseline = XGBClassifier(random_state=42, n_jobs=-1)   # all defaults, no imbalance handling
xgb_baseline.fit(X_train, y_train)                          # unscaled data, as with random forest

xgb_base_prob = xgb_baseline.predict_proba(X_test)[:, 1]
xgb_base_pred = xgb_baseline.predict(X_test)                # default 0.5 cutoff

print(f"Test PR-AUC:  {average_precision_score(y_test, xgb_base_prob):.4f}")
print(f"Precision:    {precision_score(y_test, xgb_base_pred):.3f}")
print(f"Recall:       {recall_score(y_test, xgb_base_pred):.3f}")
print(f"F1:           {f1_score(y_test, xgb_base_pred):.3f}")
print(confusion_matrix(y_test, xgb_base_pred))
```

Similar to random forest, cell 1 is a randomized search based on fixed numbers, then cell 2 chooses the cutoff, and cell 3 is the test.

```python
xgb_param_dist = {
    # Learning speed: tuned together, since a smaller learning rate needs more trees
    "learning_rate":    [0.01, 0.03, 0.05, 0.1],
    "n_estimators":     [200, 400, 800, 1200],

    # Tree complexity
    "max_depth":        [3, 4, 5, 6, 8],
    "min_child_weight": [1, 3, 5, 10],
    "gamma":            [0, 0.5, 1, 2],

    # Randomness (reduces overfitting)
    "subsample":        [0.6, 0.75, 0.9, 1.0],
    "colsample_bytree": [0.4, 0.6, 0.8, 1.0],

    # Regularization on leaf values
    "reg_lambda":       [1, 5, 10],

    # Class imbalance
    "scale_pos_weight": [1, 5, 25, 100, 600],
    "max_delta_step":   [0, 1, 5],
}

xgb_search = RandomizedSearchCV(
    XGBClassifier(random_state=42, n_jobs=-1, tree_method="hist"),   # each model uses all cores (threads)
    param_distributions=xgb_param_dist,
    n_iter=60,                      # 60 random combinations x 3 folds = 180 fits
    scoring="average_precision",    # PR-AUC
    cv=cv3,                         # same 3 folds as the random forest search, so CV scores are comparable
    n_jobs=1,                       # one fit at a time: no worker processes, no BrokenProcessPool
    random_state=42,
)
xgb_search.fit(X_train, y_train)

xgb_results = pd.DataFrame(xgb_search.cv_results_)
cols = [c for c in xgb_results.columns if c.startswith("param_")] + ["mean_test_score", "std_test_score"]
display(xgb_results.sort_values("mean_test_score", ascending=False)[cols].round(4).head(15))
print("Best:", xgb_search.best_params_, "CV PR-AUC:", round(xgb_search.best_score_, 4))
```

```python
xgb_best = xgb_search.best_estimator_
xgb_oof = cross_val_predict(xgb_best, X_train, y_train, cv=cv3, method="predict_proba")[:, 1]

prec, rec, thresholds = precision_recall_curve(y_train, xgb_oof)
f1 = 2 * prec * rec / (prec + rec + 1e-12)
best_i = np.argmax(f1[:-1])
xgb_threshold = thresholds[best_i]

print(f"Chosen threshold: {xgb_threshold:.3f}")
print(f"CV precision={prec[best_i]:.3f}  recall={rec[best_i]:.3f}  F1={f1[best_i]:.3f}")
```

```python
xgb_prob = xgb_best.predict_proba(X_test)[:, 1]
xgb_pred = (xgb_prob >= xgb_threshold).astype(int)

print(f"Test PR-AUC:  {average_precision_score(y_test, xgb_prob):.4f}")
print(f"Precision:    {precision_score(y_test, xgb_pred):.3f}")
print(f"Recall:       {recall_score(y_test, xgb_pred):.3f}")
print(f"F1:           {f1_score(y_test, xgb_pred):.3f}")
print(confusion_matrix(y_test, xgb_pred))
```

Comparison of all models:

```python
from sklearn.metrics import roc_auc_score

# (name, test probabilities, test predictions, threshold, CV PR-AUC or None for baselines)
models = [
    ("LR baseline",  baseline_prob, baseline_pred, 0.5,           None),
    ("LR tuned",     lr_prob,       lr_pred,       lr_threshold,  lr_grid.best_score_),
    ("RF baseline",  rf_base_prob,  rf_base_pred,  0.5,           None),
    ("RF tuned",     rf_prob,       rf_pred,       rf_threshold,  rf_search.best_score_),
    ("XGB baseline", xgb_base_prob, xgb_base_pred, 0.5,           None),
    ("XGB tuned",    xgb_prob,      xgb_pred,      xgb_threshold, xgb_search.best_score_),
]

rows = []
for name, prob, pred, thr, cv in models:
    tn, fp, fn, tp = confusion_matrix(y_test, pred).ravel()
    rows.append({
        "Model": name,
        "CV PR-AUC": cv,
        "Test PR-AUC": average_precision_score(y_test, prob),
        "Test ROC-AUC": roc_auc_score(y_test, prob),
        "Threshold": thr,
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1": f1_score(y_test, pred),
        "Frauds caught": f"{tp}/{tp + fn}",
        "False alarms": fp,
        "Missed frauds": fn,
    })

comparison = pd.DataFrame(rows).set_index("Model")
comparison.round(3)
```

With only 95 frauds in the test set, a few transactions can change the scores. To see how much the test results could vary by chance, we bootstrap the test set: resample it 1,000 times with replacement (frauds and non-frauds resampled separately, so each resample has 95 frauds) and recompute each metric. The models and thresholds stay fixed; nothing is retrained. The 95% confidence interval is the 2.5th to 97.5th percentile of the 1,000 scores. We also bootstrap the difference between tuned XGBoost and random forest on the same resamples.

```python
rng = np.random.default_rng(42)
n_boot = 1000
y_arr = y_test.to_numpy()
pos = np.flatnonzero(y_arr == 1)   # 95 frauds
neg = np.flatnonzero(y_arr == 0)

metric_names = ["PR-AUC", "Precision", "Recall", "F1"]
boot = {name: {m: np.empty(n_boot) for m in metric_names} for name, *_ in models}

for b in range(n_boot):
    # Stratified bootstrap: resample frauds and non-frauds separately, keeping 95 frauds each time
    idx = np.concatenate([rng.choice(pos, len(pos), replace=True),
                          rng.choice(neg, len(neg), replace=True)])
    yb = y_arr[idx]
    for name, prob, pred, _, _ in models:
        pb = np.asarray(pred)[idx]
        tp = np.sum((pb == 1) & (yb == 1))
        fp = np.sum((pb == 1) & (yb == 0))
        fn = np.sum((pb == 0) & (yb == 1))
        p = tp / (tp + fp) if tp + fp else 0.0
        r = tp / (tp + fn)
        boot[name]["PR-AUC"][b]    = average_precision_score(yb, np.asarray(prob)[idx])
        boot[name]["Precision"][b] = p
        boot[name]["Recall"][b]    = r
        boot[name]["F1"][b]        = 2 * p * r / (p + r) if p + r else 0.0

# Point estimate [95% CI] for every model and metric
ci_table = pd.DataFrame({
    m: [f"{comparison.loc[name, 'Test PR-AUC' if m == 'PR-AUC' else m]:.3f} "
        f"[{np.percentile(boot[name][m], 2.5):.3f}, {np.percentile(boot[name][m], 97.5):.3f}]"
        for name, *_ in models]
    for m in metric_names
}, index=[name for name, *_ in models])
ci_table.index.name = "Model"
display(ci_table)

# Paired differences: same resamples for both models, so the comparison is fair
pairs = [("XGB tuned", "RF tuned", "PR-AUC"),
         ("XGB tuned", "RF baseline", "PR-AUC"),
         ("XGB tuned", "RF baseline", "F1")]
for a, b_, m in pairs:
    diff = boot[a][m] - boot[b_][m]
    print(f"{a} minus {b_} ({m}): {diff.mean():+.3f}  "
          f"95% CI [{np.percentile(diff, 2.5):+.3f}, {np.percentile(diff, 97.5):+.3f}]  "
          f"{a} higher in {np.mean(diff > 0):.0%} of resamples")

# Plot: PR-AUC with 95% CI error bars
names = [name for name, *_ in models]
point = comparison.loc[names, "Test PR-AUC"].to_numpy()
lo = np.array([np.percentile(boot[n]["PR-AUC"], 2.5) for n in names])
hi = np.array([np.percentile(boot[n]["PR-AUC"], 97.5) for n in names])
plt.figure(figsize=(8, 4))
plt.errorbar(point, names, xerr=[point - lo, hi - point], fmt="o", capsize=4)
plt.gca().invert_yaxis()
plt.xlabel("Test PR-AUC (95% bootstrap CI)")
plt.title("Test PR-AUC with bootstrap confidence intervals")
plt.tight_layout()
plt.show()
```

Based on our metric, PR-AUC, the tuned XGBoost model has the highest value. It ties for the most fraud caught while having few false alarms. However, the lead is very small, and the untuned random forest has a higher test F1. The bootstrap intervals are wide (about ±0.07 PR-AUC) because the test set has only 95 frauds. Even so, the paired test shows tuned XGBoost's PR-AUC is reliably higher than both tuned random forest (+0.017, 95% CI [+0.004, +0.033]) and the random forest baseline (+0.041, 95% CI [+0.017, +0.070]). The random forest baseline's slightly higher F1 is not significant (difference −0.016, 95% CI [−0.046, +0.014]). Tuned XGBoost is therefore the best model at ranking fraud risk, while its F1 at the chosen threshold is statistically tied with random forest.


Using the tuned XGBoost model, we will use three methods to determine the most important predictors: 1) check which variable has the highest gain, or how much each feature improved the trees' splits, 2) permutation importance, shuffling one feature at a time and measuring how much PR-AUC drops, and 3) SHAP, calculating each variable's marginal contribution.

```python
gain = xgb_best.get_booster().get_score(importance_type="gain")
xgb_importance = (pd.Series(gain, name="Gain")
                    .reindex(X_train.columns, fill_value=0)   # features never used get 0
                    .sort_values(ascending=False))
xgb_importance = (xgb_importance / xgb_importance.sum()).rename("Share of total gain")

print(xgb_importance.head(15).round(3))

xgb_importance.head(15).sort_values().plot(kind="barh", figsize=(8, 6))
plt.title("XGBoost tuned: top 15 features by gain")
plt.xlabel("Share of total gain")
plt.tight_layout()
plt.show()
```

```python
from sklearn.inspection import permutation_importance

perm = permutation_importance(
    xgb_best, X_test, y_test,
    scoring="average_precision",
    n_repeats=5, random_state=42, n_jobs=1)

perm_importance = (pd.DataFrame({"PR-AUC drop": perm.importances_mean,
                                 "Std": perm.importances_std},
                                index=X_test.columns)
                     .sort_values("PR-AUC drop", ascending=False))

print(perm_importance.head(15).round(4))

top = perm_importance.head(15).iloc[::-1]
plt.figure(figsize=(8, 6))
plt.barh(top.index, top["PR-AUC drop"], xerr=top["Std"])
plt.title("XGBoost tuned: permutation importance (test PR-AUC drop)")
plt.xlabel("Drop in PR-AUC when feature is shuffled")
plt.tight_layout()
plt.show()
```

```python
import shap
import xgboost as xgb

# All test frauds + 2,000 random normal transactions
fraud_idx  = y_test[y_test == 1].index
normal_idx = y_test[y_test == 0].sample(2000, random_state=42).index
X_shap = X_test.loc[fraud_idx.union(normal_idx)]

# SHAP values (log-odds of fraud); fall back to XGBoost's own SHAP if versions clash
try:
    shap_values = shap.TreeExplainer(xgb_best)(X_shap)
except Exception as e:
    print(f"TreeExplainer failed ({e}); using XGBoost's built-in SHAP instead.")
    contribs = xgb_best.get_booster().predict(xgb.DMatrix(X_shap), pred_contribs=True)
    shap_values = shap.Explanation(values=contribs[:, :-1], base_values=contribs[:, -1],
                                   data=X_shap.values, feature_names=list(X_shap.columns))

# 1) Overall importance: mean |SHAP| per feature
shap.plots.bar(shap_values, max_display=15, show=False)
plt.title("XGBoost tuned: mean |SHAP| (top 15)")
plt.tight_layout(); plt.show()

# 2) Direction: red = high feature value, blue = low; right = pushes toward fraud
shap.plots.beeswarm(shap_values, max_display=15, show=False)
plt.title("XGBoost tuned: SHAP beeswarm")
plt.tight_layout(); plt.show()

# 3) Why one caught fraud was flagged
caught = [i for i in fraud_idx if xgb_pred[X_test.index.get_loc(i)] == 1]
i = X_shap.index.get_loc(caught[0])
shap.plots.waterfall(shap_values[i], max_display=12, show=False)
plt.title("Explanation for one caught fraud")
plt.tight_layout(); plt.show()

# Top 15 as a table
shap_rank = (pd.Series(np.abs(shap_values.values).mean(axis=0), index=X_shap.columns)
               .sort_values(ascending=False).rename("Mean |SHAP|"))
shap_rank.head(15).round(3)
```

Based on these graphs, a very negative V14 strongly signals fraud, a high V4 signals fraud, while low values of V10, V12, and V17 push towards fraud.

## Conclusion

**Goal.** Detect fraudulent credit card transactions in a highly imbalanced dataset: 283,726 transactions after removing 1,081 duplicates, of which 473 (0.167%) are fraud. Models selected by **PR-AUC** (average precision), and precision, recall and F1 also reported at a chosen threshold.

**Approach.**
- **Cleaning and features:** removed duplicates; converted `Time` to hour of day and encoded it with sine/cosine; log-transformed `Amount` (`LogAmount`), which reduced its VIF from 12.3 to 1.65.
- **Split:** stratified 80/20 split. Scaling done inside a pipeline for logistic regression only. Tree models used unscaled data.
- **Class imbalance:** handled with class weights or `scale_pos_weight` and threshold tuning, without SMOTE.
- **Tuning:** each model was tuned with cross-validation on the training set only. The decision threshold was chosen to maximize F1 on out-of-fold predictions. The test set was used once, at the end.

**Results (test set, 95 frauds):**

| Model | CV PR-AUC | Test PR-AUC | Test ROC-AUC | Precision | Recall | F1 | Caught | False alarms |
|---|---|---|---|---|---|---|---|---|
| LR baseline | – | 0.699 | 0.957 | 0.846 | 0.579 | 0.688 | 55 | 10 |
| LR tuned | 0.762 | 0.700 | 0.963 | 0.830 | 0.768 | 0.798 | 73 | 15 |
| RF baseline | – | 0.783 | 0.913 | 0.973 | 0.747 | 0.845 | 71 | 2 |
| RF tuned | 0.832 | 0.807 | 0.937 | 0.921 | 0.737 | 0.819 | 70 | 6 |
| XGB baseline | – | 0.669 | 0.902 | 0.846 | 0.695 | 0.763 | 66 | 12 |
| **XGB tuned** | **0.852** | **0.824** | **0.982** | 0.901 | **0.768** | 0.830 | **73** | 8 |

**Final model: tuned XGBoost.**
- It had the highest cross-validated PR-AUC (0.852), and the test set confirmed it with the highest test PR-AUC (0.824) and ROC-AUC (0.982).
- At a threshold of 0.558, it caught 73 of 95 frauds (77%) with only 8 false alarms out of 56,651 legitimate transactions. That is about 9 out of 10 alerts being real fraud.
- Tuning mattered most for XGBoost: test PR-AUC rose from 0.669 to 0.824.
- The lead over tuned random forest is modest, about one standard deviation of the CV scores. The untuned random forest had a slightly higher test F1 (0.845), but that difference comes from a few transactions, and the model was selected by CV rather than test results.

**Key predictors.** Gain, permutation importance and SHAP all rank **V14 and V4** as the strongest predictors, followed by V10, V12, V17 and V11. SHAP shows that **very negative values of V14, V10, V12 and V17, and high values of V4**, push predictions toward fraud. This matches the class separation found in the EDA (KS statistics of 0.76–0.84). `LogAmount` and hour of day had only small effects.

**Limitations.**
- V1–V28 are anonymized PCA components, so the predictors cannot be explained in business terms.
- The data covers only two days. A random split may overstate performance compared with an out-of-time split, where the model is tested on later transactions.
- With only 95 test frauds, a single transaction changes recall by about 1 percentage point, so small differences between models are not meaningful.
- The threshold maximizes F1, which treats a missed fraud and a false alarm as equally costly. In practice, a missed fraud usually costs more.

**Next steps.**
- Set the threshold using real costs of missed fraud vs. investigating a false alarm.
- Validate with an out-of-time split and add bootstrap confidence intervals to the test metrics.
- Retrain the final model on all of the data before deployment, and monitor it for drift.
